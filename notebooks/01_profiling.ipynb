```python
# Function for profiling data of given file
def profiling_data(file_path):
    # Reading file
    df = pd.read_csv(file_path)

    # Step 1: Basic dataset information
    print("=================Basic dataset information========================")
    # Shape
    print("Shape:", df.shape)

    # Column names
    print("Columns list:\n",df.columns)

    # Data types
    print(df.dtypes)

    # Info summary
    print(df.info())
  


    # Step 2: Missing Value Analysis
    print("\n======================Missing Value Analysis========================")
    # Missing values count
    missing_count = df.isnull().sum()

    # Missing percentage
    missing_percent = (df.isnull().sum() / len(df)) * 100

    missing_df = pd.DataFrame({
        "Missing Count": missing_count,
        "Missing %": missing_percent
    })

    print(missing_df.sort_values("Missing %", ascending=False))


    # Step 3: Duplicate Records
    print("\n========================Duplicate Records=============================")
    duplicates = df.duplicated().sum()
    print("Duplicate rows:", duplicates)
    print("Duplicate %:", duplicates / len(df) * 100)


    # Step 4. Numerical Column Statistics
    print("\n======================Numerical Column Statistics=================================")
    numeric_summary = df.describe().T
    numeric_summary["skewness"] = df.skew(numeric_only=True)
    numeric_summary["kurtosis"] = df.kurtosis(numeric_only=True)
    print(numeric_summary)


    # Step:5 Categorical Column Analysis
    print("\n===============================Categorical Column Analysis============================")
    cat_cols = df.select_dtypes(include="object").columns

    for col in cat_cols:
        print(f"\nColumn: {col}")
        print("Unique values:", df[col].nunique())
        # print(df[col].value_counts().head(10))

    # Visualization:
    # for col in cat_cols:
    #     plt.figure(figsize=(8, 4))
    #     df[col].value_counts().head(10).plot(kind='bar')
    #     plt.title(col)
    #     plt.show()

    
    # Step:6 Outlier Detection
    print("\n==================Outlier Detection=============================")
    # Using IQR:
    num_cols = df.select_dtypes(include=['int64', 'float64']).columns

    for col in num_cols:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR
        outliers = df[(df[col] < lower) | (df[col] > upper)]
        print(f"{col}: {len(outliers)} outliers")

    # Boxplot:
    # for col in num_cols:
    #     plt.figure(figsize=(8, 3))
    #     sns.boxplot(x=df[col])
    #     plt.title(col)
    #     plt.show()


    # Step:7 Distribution Analysis
    # What to profile
    # Normality
    # Skewed distributions
    # Multi-modal distributions

    # for col in num_cols:
    #     plt.figure(figsize=(8, 4))
    #     sns.histplot(df[col], kde=True)
    #     plt.title(f"Distribution of {col}")
    #     plt.show()


    # Step:8 Correlation Analysis
    print("\n===================Correlation Analysis===========================")
    # What to profile
    # Highly correlated features
    # Multicollinearity
    corr_matrix = df.corr(numeric_only=True)

    # plt.figure(figsize=(12, 8))
    # sns.heatmap(corr_matrix, annot=True, cmap="coolwarm")
    # plt.title("Correlation Matrix")
    # plt.show()

    # Highly correlated pairs:
    threshold = 0.8
    corr_pairs = (
        corr_matrix.where(
            (corr_matrix > threshold) & (corr_matrix < 1)
        ).stack()
    )

    print(corr_pairs)

    # Step:9 Data Type Consistency
    # What to profile
    # Numeric stored as string
    # Date columns
    # Mixed types

    print("\n====================Data Type Consistency=======================")
    for col in df.columns:
        print(f"{df[col].apply(type).value_counts()}\n")

    # Convert dates:
    # df['date_column'] = pd.to_datetime(df['date_column'], errors='coerce')


    # Step:10 Unique Identifier Checks
    # What to profile
    # Primary key uniqueness
    # Duplicate IDs

    # id_col = "customer_id"
    # print("Unique IDs:", df[id_col].nunique())
    # print("Total Rows:", len(df))

    # if df[id_col].nunique() != len(df):
    #     print("Duplicate IDs found")


    # Step: 11 Class Imbalance (For ML Problems)
    # What to profile
    # Target distribution
    # Minority classes

    # target = "target_column"
    # print(df[target].value_counts(normalize=True) * 100)

    # Visualization:
    # sns.countplot(x=df[target])
    # plt.title("Target Distribution")
    # plt.show()


    # Step:12 Feature Relationships
    # What to profile
    # Numeric vs numeric
    # Categorical vs numeric

    # sns.pairplot(df[num_cols])
    # plt.show()
```

### Business Meaning of below datasets
* **Customers_dataset** : Shows the data about the customers and there locality
* **Geolocation_dataset**: Contains more fine grained detail about the locality of the customers
* **Products_dataset**: Contains details about products that are available for selling
* **Sellers_dataset**: Contains details about sellers that sell the products to customers
* **Orders_dataset**: Contains details about the orders placed by customers
* **Order_items**: Shows the items placed in the cart by customers
* **Order_payments**: Contains the payment details used for payment
* **Order_reviews**: Contains the customer reviews for the orders recieved.
